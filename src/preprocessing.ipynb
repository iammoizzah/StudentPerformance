```python
# Clean each sheet individually, THEN combine them

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

print("="*80)
print("STEP 2: ADVANCED PREPROCESSING & DATA CLEANING")
print("="*80)


excel_file = '../data/raw/marks_dataset.xlsx'  
sheet_names = ['1', '2', '3', '4', '5', '6']


print("\n LOADING RAW DATA")
print("="*80)

raw_sheets = {}
for sheet in sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet)
    raw_sheets[sheet] = df
    print(f" Sheet {sheet} loaded: {df.shape}")
```

```python
# Show all columns for each sheet
for name, df in raw_sheets.items():
    print(f"\n{name}:")
    print(df.columns.tolist())
```

```python
all_cols = set().union(*[df.columns for df in raw_sheets.values()])
all_cols
```

```python
# Normalize sheets by adding missing columns
for name, df in raw_sheets.items():
    missing_cols = all_cols - set(df.columns)
    for col in missing_cols:
        df[col] = np.nan  # temporarily NaN
```

```python
all_cols = sorted(list(all_cols))

for name, df in raw_sheets.items():
    raw_sheets[name] = df[all_cols]   
```

```python
merged = pd.concat(raw_sheets.values(), ignore_index=True)

# Identify exam columns
exam_cols = [col for col in merged.columns if "midterm" in col or "final" in col]

# Drop rows where exams are missing
merged = merged.dropna(subset=exam_cols)

# Fill assignment/quiz missing values with 0
merged = merged.fillna(0)
```

```python
for col in merged.columns:
    merged[col] = pd.to_numeric(merged[col], errors='coerce')

merged = merged.fillna(0)
```

These symbols (:, -) will cause problems during:

feature selection

modeling

polynomial regression

Streamlit dashboard

So we will clean them and assign meaningful names.

```python
import re

def clean_column_names(col):
    # Remove colons and hyphens
    col = col.replace(":", "").replace("-", "")
    # Replace S_I with Mid1
    col = col.replace("SI", "Mid1")
    # Replace SII with Mid2
    col = col.replace("SII", "Mid2")
    # Replace Final with Final
    col = col.replace("Final", "Final")
    # Replace Proj with Proj
    col = col.replace("Proj", "Proj")
    # Replace As with As
    col = col.replace("As", "As")
    # Replace Qz with Qz
    col = col.replace("Qz", "Qz")
    return col

# Apply to all columns
merged.columns = [clean_column_names(col) for col in merged.columns]

merged.columns
```

```python
question_cols = [col for col in merged.columns if "_Q" in col]

merged = merged.drop(columns=question_cols)
```

```python
df = merged.copy()

df = df.rename(columns={
    "Mid1I": "Mid2",
    "Mid1I1": "Mid2_Q1",
    "Mid1I2": "Mid2_Q2",
    "Mid1I3": "Mid2_Q3"
})
```

```python
cols_to_drop = ["Mid11", "Mid12", "Mid13", "Mid2_Q1", "Mid2_Q2", "Mid2_Q3"]

df = df.drop(columns=cols_to_drop)
```

```python
df.to_csv("../data/processed/final_dataset.csv", index=False)
print("Clean dataset saved!")
```
